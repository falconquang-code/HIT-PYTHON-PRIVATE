Viết average_score(scores): list rỗng trả None; còn lại trả trung bình. Không gọi input hoặc print bên trong hàm.

**Đối chiếu:** `[8, 7, 9]` → `8.0`; `[10]` → `10`; `[]` → `None`.

```python
def average_score(scores):
    if len(scores)==0:
        return None
    else:
        return sum(scores)/len(scores)
    # TODO
print(average_score([]))
print(average_score([8,7,9]))
```

## Bài 2 · Viết hàm mới
Viết normalize_name(text) loại khoảng trắng thừa, trả chuỗi mới. Thử tên toàn khoảng trắng → chuỗi rỗng.

**Đối chiếu:** `"  Nguyen   Van An  "` → `"Nguyen Van An"`; `"   "` → `""`.

```python
def normalize_name(text):
    text=" ".join(text.title().split())
    return text
print(normalize_name("  Nguyen   Van An  "))
```

## Bài 3 · Chia chương trình thành hàm
Viết `count_passed(scores, threshold=5)` đếm điểm >= ngưỡng, không sửa list đầu vào. Viết thêm code gọi hàm này và `average_score` để in báo cáo. Thêm type hints và docstring cho hai hàm.

**Đối chiếu:** `[4, 5, 8]` với ngưỡng 5 → `2`, ngưỡng 8 → `1`; `[]` → `0`.

```python
def count_passed(scores, threshold=8):
    dem=0
    for score in scores:
        if score >= threshold:
            dem+=1
        if len(scores)==0:
            return 0
    return dem
print(count_passed([4,5,8]))
print(count_passed([]))
```

## Bài 4 · Mở rộng
Viết `calculate(a, b, operation)` cho add/subtract/multiply/divide. Chia cho 0 hoặc phép toán lạ trả None. Tự gọi hàm với các trường hợp khác nhau.

**Đối chiếu:** `calculate(6, 3, "divide")` → `2`; chia 0 hoặc phép toán lạ → `None`.

```python
# TODO: viet ham calculate va ca thu cua ban
def calculate(a, b, operation):
    if operation =="add":
        return a + b
    elif operation=="subtract":
        return a - b
    elif operation=="multiply":
        return a * b
    elif operation=="divide":
        return a / b
    else:
        return None
calculate(6, 3, "divide")
```
