## Bài 2 · Thống kê list
Cho điểm `[8,4,9,5]`. Tính trung bình, cao nhất và số điểm >=5 bằng vòng lặp. Giữ nguyên list gốc.

**Đối chiếu:** Trung bình `6.5`, cao nhất `9`, số điểm đạt `3`.

```python
scores = [8, 4, 9, 5]
average = sum(scores) / len(scores)
highest = max(scores)
passed_count = [i for i in scores if i >=5]
# TODO: tinh ket qua; khong gan truc tiep dap an
print(average)
print(highest)
print(passed_count)
```

## Bài 3 · Viết mới lọc danh sách
Cho tên có khoảng trắng/trùng sau chuẩn hóa. Tạo unique_names theo thứ tự gặp, bỏ tên rỗng và tên trùng. Chỉ dùng string/list/loop/if; chưa cần set.

**Đối chiếu:** `["Lan", "Minh", "An"]`, giữ thứ tự gặp và bỏ tên rỗng.

```python
raw_names = [" Lan ", "Minh", "Lan", "  ", " An  "]
unique_names = []
# TODO: tao list moi, khong sua raw_names
for name in raw_names:
    clean_name= name.strip()
    if clean_name !="" and clean_name not in unique_names:
        unique_names.append(clean_name)
print(unique_names)
```

## Bài 5 · Set và tuple
Tạo set các môn duy nhất; tạo tuple tên ba môn đã sắp xếp theo chữ cái.

**Đối chiếu:** Set gồm Git, Python, Web; tuple `("Git", "Python", "Web")`.

```python
subjects = ["Python", "Web", "Python", "Git", "Web"]
unique_subjects =set(subjects)
subject_tuple = tuple(sorted(unique_subjects))
print(unique_subjects)
print(subject_tuple)
```

## Bài 6 · Thống kê bản ghi
Tự tạo dict đếm học viên theo group từ list dưới. Không gán đáp án trực tiếp; dùng loop và get.

**Đối chiếu:** `{"Python": 2, "Web": 1}`; list rỗng cho `{}`.

```python
students = [
    {"name": "Lan", "group": "Python"},
    {"name": "Minh", "group": "Web"},
    {"name": "An", "group": "Python"},
]
counts = {}
for student in students:
    group=student["group"]
    counts[group]=counts.get(group,0)+1
# TODO: tong hop so hoc vien theo group
print(counts)
```
